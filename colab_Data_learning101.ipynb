```python
# 導入 pandas 庫，用於數據處理和分析
import pandas as pd

# 從 Google Drive 讀取 CSV 檔案到 DataFrame 中
df = pd.read_csv("/content/drive/MyDrive/python 商業分析/coffee_shop.csv")

# 顯示 DataFrame 的類型，確認它是一個 pandas DataFrame 物件
type(df)
```

```python
# 顯示 DataFrame 的前幾行，快速預覽數據內容
df.head()
```

```python
# 選取並顯示 DataFrame 中指定欄位的數據 ('Product', 'Category', 'Price', 'Quantity')
df[['Product', 'Category', 'Price', 'Quantity']]
```

```python
# 創建一個新欄位 'Totalprice'，計算每個商品的總價格 (Price * Quantity)
df['Totalprice'] = df['Price'] * df['Quantity']
# 顯示更新後 DataFrame 的前幾行，包含新的 'Totalprice' 欄位
df.head()
```

```python
# 根據 'Product' 欄位對 DataFrame 進行分組
group = df.groupby('Product')
# 顯示分組物件的類型及其包含的組數
print(type(group)) , len(group)

# 以下是被註釋掉的程式碼，用於遍歷並印出每個產品組的項目數量，目前未執行
#print("\n每個產品的項目數量：")
#for product_name, group_data in group:
 #   print(f"產品: {product_name}, 數量: {len(group_data)}")
```

```python
# 遍歷根據 'Product' 欄位分組後的每個產品群組
for product_name , group_data in df.groupby('Product'):
  # 顯示產品名稱以及該產品群組包含的數據筆數
  print("產品名稱" , product_name, "共" ,len(group_data),"筆")
  # 顯示當前產品群組的所有數據
  display (group_data)
  print()
```

```python
# 從依 'Member' 欄位分組的數據中，取出 'Yes' 成員的所有數據
group_member_Yes = df.groupby('Member').get_group('Yes')
# 顯示取出數據的類型及其筆數
print(type(group_member_Yes),len(group_member_Yes))
# 顯示取出數據的前幾行
display(group_member_Yes.head())
print()
```

```python
# 從依 'Product' 欄位分組的數據中，取出 'Latte' 產品的所有數據
latte_costermer = df.groupby('Product').get_group('Latte')
# 顯示取出數據的類型及其筆數
print(type(latte_costermer),len(latte_costermer))
# 顯示取出數據的所有內容
display(latte_costermer)
print()
```

```python
# 使用布林索引篩選出 'Product' 為 'Latte' 且 'Member' 為 'Yes' 的數據
latte_costermer_member = df[(df["Product"]=="Latte") & (df["Member"]=="Yes")]
# 顯示篩選後數據的類型及其筆數
print(type(latte_costermer_member),len(latte_costermer_member))
# 顯示篩選後數據的所有內容
display(latte_costermer_member)
print()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 設定圖表風格
sns.set_style("whitegrid")

# 計算每個產品的總銷售數量 (若尚未計算，則在此處重新計算)
product_sales = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)

# 繪製長條圖
# 確保 product_sales 在此儲存格中定義
product_sales = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(
    x=product_sales.index,
    y=product_sales.values,
    palette='coolwarm'
    )
plt.title('Product Sales Quantity')
plt.xlabel('Product')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45, ha='right')
plt.savefig('product_sales.png') # before show fig

plt.tight_layout()
plt.show()
```

```python
# 確保 'Totalprice' 欄位存在，如果不存在則重新計算
if 'Totalprice' not in df.columns:
    df['Totalprice'] = df['Price'] * df['Quantity']

# 計算每個類別的總銷售額
category_sales = df.groupby('Category')['Totalprice'].sum()

# 繪製圓餅圖
plt.figure(figsize=(8, 8))
plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel'), # 修正：在這裡添加逗號
    textprops={'fontsize': 20} # 增加百分比文字大小
    )

plt.title('Sales Proportion by Category',fontsize=16 )
plt.ylabel('') # 移除Y軸標籤以避免與標題重疊
#plt.axis('equal') # 確保圓餅圖是圓形的
plt.tight_layout()
plt.savefig('sales_proportion_by_category.png')
plt.show()
```

```python
member_category_sales = df.groupby(['Member', 'Category'])['Totalprice'].sum().unstack()
print("unstock")
display(member_category_sales)
```

```python
# 繪製堆疊長條圖或分組長條圖
# 將 DataFrame 轉換為適合繪圖的格式
member_category_sales_melted = member_category_sales.reset_index().melt(
    'Member', var_name='Category', value_name='TotalSales')
```

```python
plt.figure(figsize=(12, 7))
sns.barplot(
    x='Category',
    y='TotalSales',
    hue='Member',
    data=member_category_sales_melted,
    palette='viridis'
)
plt.title('Total Sales by Category and Member Type', fontsize=16)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Total Sales Amount', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Member')
plt.savefig('member_category_preference.png') # 儲存圖片
plt.show()
```

```python
member_Product_sales = df.groupby(['Member', 'Product'])['Totalprice'].sum().unstack()
print("unstock")
display(member_Product_sales)
# 繪製堆疊長條圖或分組長條圖
# 將 DataFrame 轉換為適合繪圖的格式
member_Product_sales_melted = member_Product_sales.reset_index().melt(
    'Member', var_name='Product', value_name='TotalSales')
display(member_Product_sales_melted)
plt.figure(figsize=(12, 7))
sns.barplot(
    x='Product',
    y='TotalSales',
    hue='Member',
    data=member_Product_sales_melted,
    palette='viridis'
)
plt.title('Total Sales by Product and Member Type', fontsize=16)
plt.xlabel('Product', fontsize=12)
plt.ylabel('Total Sales Amount', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Member')
plt.savefig('member_Product_preference.png') # 儲存圖片
plt.show()
```
